### 2. Indexing: one value out of the list

Square brackets after the list name pull out a single value. **Python counts from
0**, which is actually convenient here: index 0 is the hour starting 00:00, index 6
is 06:00, index 13 is 13:00 — the index *is* the hour.

```python
print(f"Midnight: {pm_hourly[0]} µg/m³")
print(f"6 a.m.:   {pm_hourly[6]} µg/m³")
```

Negative indexes count from the end — `-1` is the last value (23:00):

```python
print(f"11 p.m.:  {pm_hourly[-1]} µg/m³")
```

### 5. Loops: do something with every value

A `for` loop visits each value in turn. The indented lines run once per value.
Here's the classic **counting pattern** — how many hours exceeded the WHO 24-hour
guideline level?

```python
WHO_DAILY = 15  # µg/m³

hours_above = 0
for value in pm_hourly:
    if value > WHO_DAILY:
        hours_above = hours_above + 1

print(f"{hours_above} of {len(pm_hourly)} hours above {WHO_DAILY} µg/m³")
```

**Every single hour** of this day was above the WHO guideline level — a normal
Harmattan-season day at a busy Lagos transport hub.

The same pattern with a running total is how a mean is computed "by hand" —
`sum()` is doing exactly this for you:

```python
total = 0.0
for value in pm_hourly:
    total = total + value

print(f"Mean via loop: {total / len(pm_hourly):.1f} µg/m³")
```

### 6. Dictionaries: labelled values

A list is ordered but unlabelled. A **dictionary** (dict) stores values under

```python
station = {
    "name": "Oshodi Bus Terminal",
    "city": "Lagos",
    "latitude": 6.5567,
    "longitude": 3.3523,
    "sensor_type": "low-cost optical",
}
print(station["name"])
print(f"Located at {station['latitude']}, {station['longitude']}")
```

You can add or change entries by assigning to a key:

```python
station["operator"] = "AirGradient network"
print(station)
```

### 7. The punchline: a table is a list of dictionaries

Put the two ideas together — a **list** of **dicts**, one dict per measurement —
and you have *records*: rows, each with named columns. This is what a data table
fundamentally is (and it's exactly the shape web APIs use, as you'll see in the
bonus notebook):

```python
measurements = [
    {"hour": 7, "station": "Oshodi Bus Terminal", "pm25_value": pm_hourly[7]},
    {"hour": 8, "station": "Oshodi Bus Terminal", "pm25_value": pm_hourly[8]},
    {"hour": 9, "station": "Oshodi Bus Terminal", "pm25_value": pm_hourly[9]},
]

for row in measurements:
    print(f"{row['hour']:>2d}:00  {row['station']}  {row['pm25_value']} µg/m³")
```

Managing thousands of rows this way — looping for every mean, every filter — gets
old fast. The **pandas** library wraps this exact shape in an object called a
`DataFrame` that can compute means, filter rows, and resample time series in one
line each.

### 8. Recap

* **Lists** hold ordered values; index from 0 (`pm_hourly[6]` = 06:00), slice with
  `[start:stop]` (stop excluded).
* `len`, `min`, `max`, `sum` consume whole lists.
* **Loops** apply logic to every value — counting and totalling are the two
  patterns you'll reuse forever.
* **Dicts** label values by key; a **list of dicts** is a table.
