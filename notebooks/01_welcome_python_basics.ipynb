### 2. Variables: giving values a name

Analysis means keeping track of values — a concentration, a station name, a
guideline. A **variable** gives a value a name you can reuse. The `=` sign means
"store the value on the right under the name on the left."

```python
pm25 = 87.4  # a PM2.5 concentration in µg/m³ (micrograms per cubic metre)
print(pm25)
```

Variables hold text too — just add quotes:

```python
station = "Lagos (US Consulate)"
print(station)
```

### 3. Writing sentences with numbers: f-strings

Analysts constantly write sentences with numbers in them. An **f-string** is a string
with an `f` in front, where anything inside `{curly braces}` gets replaced by its
value:

```python
print(f"PM2.5 at {station}: {pm25} µg/m³")
```

Adding `:.1f` inside the braces rounds a number to 1 decimal place — useful because
instruments report more digits than anyone needs:

```python
pm25_precise = 87.4321
print(f"Rounded: {pm25_precise:.1f} µg/m³")
```

### 5. Arithmetic

Python is a good calculator: `+`, `-`, `*` (multiply), `/` (divide). Suppose a
station measured 95.2 µg/m³ at 09:00 and 88.6 µg/m³ at 10:00. The two-hour average:

```python
pm_9am = 95.2
pm_10am = 88.6
average = (pm_9am + pm_10am) / 2
print(f"Two-hour average: {average:.1f} µg/m³")
```

A calculation you'll use constantly: **how many times higher** one value is than
another. During the Harmattan (the dry, dusty season, roughly December–February),
Lagos PM2.5 is far higher than in the rainy season:

```python
harmattan_pm25 = 110.5  # a typical December value
rainy_pm25 = 38.2       # a typical August value
percent_higher = (harmattan_pm25 - rainy_pm25) / rainy_pm25 * 100
print(f"Harmattan is {percent_higher:.0f}% higher than the rainy season")
```

### 6. Comparisons and decisions

The WHO 2021 guideline for 24-hour average PM2.5 is **15 µg/m³**. A comparison asks
a yes/no question and returns `True` or `False`:

```python
WHO_DAILY = 15  # µg/m³, WHO 2021 guideline for 24-hour average PM2.5
today = 87.4
print(today > WHO_DAILY)
```

`if` / `else` lets the notebook *act* on that answer. Note the indentation — the
indented lines belong to their `if` or `else` branch:

```python
if today > WHO_DAILY:
    print(f"Above the WHO daily guideline ({today / WHO_DAILY:.1f}× the guideline)")
else:
    print("Within the WHO daily guideline")
```

With `elif` ("else if") you can check several ranges in order — like a simplified
air-quality advisory. Python works down the list and takes the **first** branch that
matches:

```python
value = 87.4
if value < 15:
    advisory = "good"
elif value < 50:
    advisory = "moderate"
else:
    advisory = "unhealthy"
print(f"{value} µg/m³ → {advisory}")
```
